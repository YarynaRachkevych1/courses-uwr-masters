```python
import numpy as np
import plotly.express as px
import pandas as pd
```

```python
data = np.random.randn(1_000_000)

px.histogram(
    x=data,
    nbins=100,
    title="Histogram rozkładu normalnego N(0, 1)",
    labels={
        "x": "Wartość",
        "y": "Liczba obserwacji"
    }
)
```

```python
columns = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width",
    "species"
]

iris = pd.read_csv("iris.data", names=columns)
```

```python
s = 0.25

iris_noisy = iris.copy()

iris_noisy["sepal_length"] = (
    iris["sepal_length"] + np.random.normal(0, s, len(iris))
)

iris_noisy["sepal_width"] = (
    iris["sepal_width"] + np.random.normal(0, s, len(iris))
)
```

```python
plot_sepal = px.scatter(iris, x="sepal_length", y="sepal_width", color="species", size="petal_length")
plot_sepal.show()
```

```python
plot_sepal_noisy = px.scatter(iris_noisy, x="sepal_length", y="sepal_width", color="species", size="petal_length")
plot_sepal_noisy.show()
```

```python
iris_setosa_noisy = iris_noisy[(iris_noisy["species"] == "Iris-setosa")]
print(iris_setosa_noisy["petal_length"].max())
```

Po kilkukrotnym dodaniu szumu o odchyleniu standardowym s = 0.25 (i większym) wcześniejsza reguła klasyfikacji nadal działa:
kwiat należy do gatunku "iris-setosa", jeśli petal_length < 2. Wynika to z faktu, że szum został dodany jedynie do 
wymiarów działki kielicha (sepal_length i sepal_width), a nie do długości płatka (petal_length). 
Na wykresie punkty reprezentujące ten gatunek są jednak bardziej rozproszone i nie tworzą już tak zwartej grupy 
jak w przypadku danych oryginalnych.
