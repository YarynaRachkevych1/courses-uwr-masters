```python
import numpy as np
import plotly.graph_objects as go
```

```python
n = 1_000_000
data = np.random.randn(n, 2)

bins = 50
hist_range = [[-4, 4], [-4, 4]]

hist, x_edges, y_edges = np.histogram2d(
    data[:, 0],
    data[:, 1],
    bins=bins,
    range=hist_range,
    density=True
)

x = (x_edges[:-1] + x_edges[1:]) / 2
y = (y_edges[:-1] + y_edges[1:]) / 2

X, Y = np.meshgrid(x, y)
Z_hist = hist.T

Z_pdf = (1 / (2 * np.pi)) * np.exp(-(X**2 + Y**2) / 2)
fig = go.Figure()

# powierzchnia histogramu
fig.add_trace(go.Surface(
    x=X,
    y=Y,
    z=Z_hist,
    colorscale=[[0, 'rgba(30, 144, 255, 0.25)'], [1, 'rgba(30, 144, 255, 0.9)']],
    showscale=False
))

# teoretyczna gęstość
fig.add_trace(go.Surface(
    x=X,
    y=Y,
    z=Z_pdf,
    colorscale=[[0, 'rgba(255, 0, 0, 0.15)'], [1, 'rgba(255, 0, 0, 0.8)']],
    showscale=False
))

fig.update_layout(
    title="Histogram i gęstość dwuwymiarowego rozkładu normalnego",
    scene=dict(
        xaxis_title="X",
        yaxis_title="Y",
        zaxis_title="Gęstość",
        xaxis=dict(range=[-4, 4]),
        yaxis=dict(range=[-4, 4]),
        zaxis=dict(range=[0, 0.18])
    ),
    width=900,
    height=700
)

fig.show()
```
